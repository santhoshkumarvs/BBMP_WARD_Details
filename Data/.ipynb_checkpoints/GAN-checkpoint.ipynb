```python
# Define the generator network
generator = Sequential()
generator.add(Dense(256, input_dim=100))
generator.add(LeakyReLU(0.2))
generator.add(BatchNormalization(momentum=0.8))
generator.add(Dense(512))
generator.add(LeakyReLU(0.2))
generator.add(BatchNormalization(momentum=0.8))
generator.add(Dense(data_dim, activation='tanh'))
generator.compile(loss='binary_crossentropy', optimizer=Adam(0.0002, 0.5))
```

```python
# Define the discriminator network
discriminator = Sequential()
discriminator.add(Dense(512, input_dim=data_dim))
discriminator.add(LeakyReLU(0.2))
discriminator.add(BatchNormalization(momentum=0.8))
discriminator.add(Dense(256))
discriminator.add(LeakyReLU(0.2))
discriminator.add(BatchNormalization(momentum=0.8))
discriminator.add(Dense(1, activation='sigmoid'))
discriminator.compile(loss='binary_crossentropy', optimizer=Adam(0.0002, 0.5))
```

```python
# Define the combined generator and discriminator model
z = Input(shape=(100,))
x_gen = generator(z)
discriminator.trainable = False
validity = discriminator(x_gen)
combined = Model(z, validity)
combined.compile(loss='binary_crossentropy', optimizer=Adam(0.0002, 0.5))
```

```python
# Train the GAN
batch_size = 32
valid = np.ones((batch_size, 1))
fake = np.zeros((batch_size, 1))
for epoch in range(10000):
    # Train the discriminator
    d_loss_real = discriminator.train_on_batch(np.random.normal(size=(batch_size, data_dim)), valid)
    d_loss_fake = discriminator.train_on_batch(generator.predict(np.random.normal(size=(batch_size, 100))), fake)
    d_loss = 0.5 * np.add(d_loss_real, d_loss_fake)
```
