```python
import pandas as pd
df = pd.read_csv('eco.csv', header=0)

```

```python

print(df.columns)


```

```python
df.columns
# Output: ['male_population', 'female_population', 'ward_Name', 'distance']
```

```python
# Create a new column 'total_population'
df['total_population'] = df['male_population'] + df['female_population']
```

```python
print(df)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(df[['male_population', 'female_population']], df['total_population'], test_size=0.2, random_state=42)
```

```python
# Create a linear regression model
model = LinearRegression()
```

```python
# Train the model on the training data
model.fit(X_train, y_train)

```

```python
# Make predictions on the testing data
y_pred = model.predict(X_test)

```

```python
# Print the predicted values
print(y_pred)

```
